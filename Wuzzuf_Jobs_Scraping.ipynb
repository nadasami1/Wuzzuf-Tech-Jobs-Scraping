## Wuzzuf Tech Jobs Web Scraping

### 📌 Project Overview

This project uses **Python, Selenium, and Pandas** to scrape technology-related job opportunities from Wuzzuf.

The scraper searches for different technology positions and extracts useful information such as job title, company, location, job type, required experience, and job URL.

The collected data is then stored in a structured Pandas DataFrame and exported to a CSV file for further analysis.


###  🎯 Project Objectives

* Automate the job searching process using Selenium.
* Scrape technology-related job listings from Wuzzuf.
* Collect relevant information about each job.
* Store the scraped data in a structured format.
* Export the collected data to a CSV file.
* Prepare the dataset for future analysis and visualization.


```python
from selenium import webdriver 
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys 
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time 
import pandas as pd
options = Options() 
options.add_argument("--disable-blinkfeatures=AutomationControlled") 
options.add_experimental_option("excludeSwitches", ["enableautomation"]) 
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless") # run in background options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new") 
options.add_argument("--window-size=1920,1080") 
options.add_argument("--disable-gpu") 
options.add_argument("--no-sandbox") 
options.add_argument("--disable-dev-shmusage") 
driver = webdriver.Chrome(options=options)

Positions = ["Data Engineer", "Data Science", "AI Developer", "Machine Learning Engineer"]

driver.get("https://wuzzuf.net/jobs/egypt")
wait = WebDriverWait(driver, 15)

all_jobs = []

for position in Positions:

    driver.find_element(By.NAME ,"q").clear()
    
    driver.find_element(By.NAME ,"q").send_keys(position)

    driver.find_element(By.XPATH,"//button[@type='submit']").click()

    time.sleep(5)
    driver.refresh()

    wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "h2 a")))

    for i in range (2):
       time.sleep(5)
       job_cards = driver.find_elements(By.CSS_SELECTOR,".css-ghe2tq")

       for job in job_cards:
          title = job.find_element(By.CLASS_NAME,"css-o171kl").text.strip()
          company = job.find_element(By.CLASS_NAME,"css-ipsyv7").text.strip()
          location = job.find_element(By.CLASS_NAME,"css-16x61xq").text.strip()
          job_type = job.find_element(By.CLASS_NAME,"css-a85cz4").text.strip()
          try:
            exp = job.find_element(By.XPATH, ".//span[contains(., 'Yrs of Exp')]").text
            exp = exp.lstrip(". ").strip()
          except Exception:
            exp = None
          Url = job.find_element(By.CLASS_NAME,"css-o171kl").get_attribute("href")

          all_jobs.append({
              "position" : position,
              "title": title,
              "company": company,
              "location": location,
              "job_type": job_type,
              "experience": exp,
              "url": Url
         })
       time.sleep(5)
       try:
        next_button = driver.find_element(By.CSS_SELECTOR, "button[class='css-kefogq ezfki8j0']")
        if next_button.is_enabled():
            next_button.click()
        else:
            break 
       except :
        print(f"An error occurred while trying to click the next button for {position}.")

jobs_df = pd.DataFrame(all_jobs)
```
